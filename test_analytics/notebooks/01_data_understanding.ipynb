# 2. Дослідження структури та взаємозв'язків даних (Data Understanding)

```python
# Імпортуємо необхідні бібліотеки
import pandas as pd
from pathlib import Path
from IPython.display import display
```

```python
# Визначаємо шлях до директорії з вихідними даними
current_dir = Path.cwd()

possible_paths = [
    current_dir / "data" / "simulated",
    current_dir.parent / "data" / "simulated",
    current_dir.parent.parent / "data" / "simulated"
]

data_path = next(
    (
        path for path in possible_paths
        if path.is_dir()
    ),
    None
)

if data_path is None:
    raise FileNotFoundError(
        "Директорію data/simulated не знайдено. "
        "Перевірте розташування вихідних CSV-файлів."
    )

print("Шлях до вихідних даних:", data_path.resolve())
```

```python
# Завантажуємо вихідні CSV-файли в окремі DataFrame

csv_files = sorted(data_path.glob("*.csv"))

if not csv_files:
    raise FileNotFoundError(
        "У директорії data/simulated відсутні CSV-файли."
    )

datasets = {
    file.stem: pd.read_csv(file)
    for file in csv_files
}

print(f"Кількість завантажених наборів даних: {len(datasets)}")
```

```python
# Формуємо загальну інформацію про кожен набір даних

dataset_summary = pd.DataFrame([
    {
        "dataset": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "total_cells": df.size
    }
    for name, df in datasets.items()
])

display(dataset_summary)

print("Загальна кількість записів:", dataset_summary["rows"].sum())
print("Кількість унікальних наборів даних:", len(datasets))
```

```python
# Отримуємо описову статистику для кожного набору даних

for name, df in datasets.items():

    print(f"\n{'=' * 70}")
    print(f"Набір даних: {name}")

    display(df.describe(include="all").T)
```

```python
# Формуємо профіль стовпців кожного набору даних

profile_data = []

for name, df in datasets.items():

    for column in df.columns:

        profile_data.append({
            "dataset": name,
            "column": column,
            "dtype": str(df[column].dtype),
            "total_rows": len(df),
            "non_null": df[column].count(),
            "missing": df[column].isna().sum(),
            "missing_pct": df[column].isna().mean() * 100,
            "unique": df[column].nunique()
        })

data_profile = pd.DataFrame(profile_data)

display(data_profile.round(2))
```

```python
# Виділяємо колонки, у яких відсутні всі значення

fully_missing = data_profile[
    data_profile["missing"] == data_profile["total_rows"]
]

print("Колонки без жодного заповненого значення:")
display(fully_missing)
```

```python
# Виділяємо колонки з частково пропущеними значеннями

partially_missing = data_profile[
    (data_profile["missing"] > 0) &
    (data_profile["missing"] < data_profile["total_rows"])
].sort_values(
    ["dataset", "missing_pct"],
    ascending=[True, False]
)

print("Колонки з частково пропущеними значеннями:")
display(partially_missing)
```

```python
# Визначаємо набори даних для аналізу структури подій

datasets_to_check = [
    "06_guest_swipe_session",
    "06_authenticated_swipe_session",
    "08_guest_results_winners",
    "08_authenticated_results_winners"
]

columns_to_check = [
    "user_id",
    "participant_id",
    "room_id",
    "archive_session_id",
    "title_id",
    "room_title_id",
    "interaction_method",
    "is_liked",
    "started_at",
    "completed_at",
    "card_position",
    "title",
    "likes_count"
]

for name in datasets_to_check:

    if name not in datasets:
        print(f"Набір даних не знайдено: {name}")
        continue

    df = datasets[name]

    # Враховуємо лише наявні колонки
    available_columns = [
        col for col in columns_to_check
        if col in df.columns
    ]

    # Рахуємо кількість записів за типом події
    event_counts = df["event_name"].value_counts()

    # Розраховуємо відсоток пропущених значень за типом події
    missing_by_event = (
        df.groupby("event_name")[available_columns]
        .agg(lambda x: x.isna().mean() * 100)
        .round(2)
    )

    event_analysis = event_counts.to_frame(
        "event_count"
    ).join(missing_by_event)

    print(f"\n{'=' * 80}")
    print(f"Набір даних: {name}")

    display(event_analysis)
```

```python
# Аналізуємо представленість ідентифікаторів у наборах даних

id_columns = [
    "event_id",
    "page_view_id",
    "visitor_id",
    "user_id",
    "participant_id",
    "room_id",
    "archive_session_id",
    "title_id",
    "room_title_id"
]

id_summary = []

for name, df in datasets.items():

    for column in id_columns:

        if column in df.columns:

            non_null = df[column].count()
            unique_values = df[column].nunique()

            id_summary.append({
                "dataset": name,
                "identifier": column,
                "total_rows": len(df),
                "non_null": non_null,
                "unique_values": unique_values,
                "repeated_occurrences": non_null - unique_values
            })

id_summary_df = pd.DataFrame(id_summary)

display(id_summary_df)
```

```python
# Об'єднуємо інформацію про відвідувачів та перегляди сторінок

tracking_columns = [
    "event_id",
    "event_timestamp",
    "event_name",
    "visitor_id",
    "user_id",
    "user_type",
    "page_view_id"
]

tracking_data = pd.concat(
    [
        df[tracking_columns].assign(dataset=name)
        for name, df in datasets.items()
    ],
    ignore_index=True
)

print("Загальна кількість записів:", len(tracking_data))
```

```python
# Досліджуємо зв'язок між visitor_id та іншими ідентифікаторами

visitor_mapping = (
    tracking_data
    .dropna(subset=["visitor_id"])
    .groupby("visitor_id")
    .agg(
        events_count=("event_id", "count"),
        unique_users=("user_id", "nunique"),
        unique_user_types=("user_type", "nunique"),
        page_views=("page_view_id", "nunique")
    )
    .reset_index()
)

display(visitor_mapping.head(10))
```

```python
# Досліджуємо кількість подій у межах одного перегляду сторінки

page_view_mapping = (
    tracking_data
    .dropna(subset=["page_view_id"])
    .groupby("page_view_id")
    .agg(
        events_count=("event_id", "count"),
        unique_visitors=("visitor_id", "nunique"),
        unique_users=("user_id", "nunique"),
        datasets=("dataset", "nunique")
    )
    .reset_index()
)

print("Унікальних visitor_id:", tracking_data["visitor_id"].nunique())
print("Унікальних page_view_id:", tracking_data["page_view_id"].nunique())
print("Подій без visitor_id:", tracking_data["visitor_id"].isna().sum())
print("Подій без page_view_id:", tracking_data["page_view_id"].isna().sum())

print("\nPage View ID, пов'язані з кількома відвідувачами:")

display(
    page_view_mapping[
        page_view_mapping["unique_visitors"] > 1
    ]
)

print("\nРозподіл кількості подій на один перегляд сторінки:")

display(
    page_view_mapping["events_count"].describe().round(2)
)
```

```python
# Виділяємо відвідувачів із кількома типами користувачів або user_id

multi_identity_visitors = visitor_mapping[
    (visitor_mapping["unique_users"] > 1) |
    (visitor_mapping["unique_user_types"] > 1)
]["visitor_id"]

visitor_details = (
    tracking_data[
        tracking_data["visitor_id"].isin(multi_identity_visitors)
    ]
    .groupby(["visitor_id", "user_type"], dropna=False)
    .agg(
        events_count=("event_id", "count"),
        unique_users=("user_id", "nunique"),
        user_ids=(
            "user_id",
            lambda x: sorted(x.dropna().astype(str).unique())
        ),
        datasets=("dataset", "nunique")
    )
    .reset_index()
)

display(visitor_details)
```

```python
# Формуємо єдиний набір даних для огляду сутностей

all_events = pd.concat(
    [
        df.assign(dataset=name)
        for name, df in datasets.items()
    ],
    ignore_index=True
)
```

```python
# Аналізуємо кількість унікальних сутностей у всіх наборах даних

entity_columns = [
    "visitor_id",
    "user_id",
    "participant_id",
    "room_id",
    "archive_session_id",
    "title_id",
    "room_title_id"
]

entity_overview = []

for column in entity_columns:

    if column in all_events.columns:

        entity_overview.append({
            "entity": column,
            "non_null_records": all_events[column].count(),
            "unique_entities": all_events[column].nunique()
        })

entity_overview_df = pd.DataFrame(entity_overview)

display(entity_overview_df)
```

```python
# Формуємо реєстр подій з усіх наборів даних

event_registry = (
    all_events
    .groupby(
        ["dataset", "event_name", "event_value_type"],
        dropna=False
    )
    .agg(
        records=("event_id", "count"),
        unique_values=("event_value", "nunique")
    )
    .reset_index()
    .sort_values(["dataset", "event_name"])
)

display(event_registry)
```

```python
# Аналізуємо типи значень однакових подій у різних наборах даних

event_type_consistency = (
    event_registry
    .groupby("event_name")
    .agg(
        datasets_count=("dataset", "nunique"),
        value_types=(
            "event_value_type",
            lambda x: sorted(x.dropna().unique().tolist())
        )
    )
    .reset_index()
)

display(
    event_type_consistency[
        event_type_consistency["datasets_count"] > 1
    ]
)
```

```python
# Визначаємо приклади значень для основних типів подій

events_to_inspect = {
    "01_landing_page": [
        "create_room_submitted",
        "guest_start_clicked"
    ],
    "06_guest_swipe_session": [
        "swipe_right",
        "swipe_left",
        "Like_button",
        "dislike_button",
        "swipe_session_started",
        "swipe_session_completed"
    ],
    "06_authenticated_swipe_session": [
        "movie_liked",
        "movie_disliked",
        "swipe_session_started",
        "swipe_session_completed"
    ],
    "08_guest_results_winners": [
        "movie_won"
    ],
    "08_authenticated_results_winners": [
        "movie_won"
    ]
}

for dataset_name, event_names in events_to_inspect.items():

    if dataset_name not in datasets:
        continue

    df = datasets[dataset_name]

    selected_events = df[
        df["event_name"].isin(event_names)
    ]

    value_summary = (
        selected_events
        .groupby(
            ["event_name", "event_value_type"],
            dropna=False
        )
        .agg(
            records=("event_id", "count"),
            unique_values=("event_value", "nunique"),
            examples=(
                "event_value",
                lambda x: (
                    x.dropna()
                    .astype(str)
                    .drop_duplicates()
                    .head(5)
                    .tolist()
                )
            )
        )
        .reset_index()
    )

    print(f"\n{'=' * 70}")
    print(dataset_name)

    display(value_summary)
```
